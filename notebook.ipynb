```python
from ucimlrepo import fetch_ucirepo
import pandas as pd

# Descargar dataset
dataset = fetch_ucirepo(id=350)

# Separar variables
X = dataset.data.features
y = dataset.data.targets

# Unir los datos
df = pd.concat([X, y], axis=1)

# Guardar CSV
df.to_csv("credit_risk.csv", index=False)

# Inspeccionar
print(df.shape)
print(df.head())
```

```python
# Dimensiones
print("Dimensiones:", df.shape)

# Primeras observaciones
display(df.head())

# Información general
df.info()

# Estadísticas descriptivas
display(df.describe())
```

```python
# Corregir nombres 
column_names = [
    "LIMIT_BAL",
    "SEX",
    "EDUCATION",
    "MARRIAGE",
    "AGE",
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6",
    "BILL_AMT1",
    "BILL_AMT2",
    "BILL_AMT3",
    "BILL_AMT4",
    "BILL_AMT5",
    "BILL_AMT6",
    "PAY_AMT1",
    "PAY_AMT2",
    "PAY_AMT3",
    "PAY_AMT4",
    "PAY_AMT5",
    "PAY_AMT6",
    "DEFAULT"
]

df.columns = column_names
```

```python
print(df.columns.tolist())
```

```python
# 1. Valores faltantes
print("Valores faltantes:")
print(df.isnull().sum())

# 2. Registros duplicados
print("Duplicados:", df.duplicated().sum())

# 3. Valores únicos por variable
print("Valores únicos:")
print(df.nunique())

# 4. Estadísticas descriptivas
display(df.describe().T)
```

```python
# Distribución de la variable objetivo
print(df["DEFAULT"].value_counts())

# Porcentaje de incumplimiento
default_rate = df["DEFAULT"].mean() * 100

print(f"Tasa de incumplimiento: {default_rate:.2f}%")
```

```python
categorical_cols = [
    "SEX", "EDUCATION", "MARRIAGE",
    "PAY_0", "PAY_2", "PAY_3",
    "PAY_4", "PAY_5", "PAY_6"
]

for col in categorical_cols:
    print(f"\nVariable: {col}")
    print(df[col].value_counts().sort_index())
```

```python
duplicates = df[df.duplicated(keep=False)]

print("Filas involucradas:", len(duplicates))
display(duplicates.head(10))
```

```python
# Agrupar categorías desconocidas de educación
df["EDUCATION"] = df["EDUCATION"].replace({
    0: 4,
    5: 4,
    6: 4
})

# Agrupar categoría desconocida de estado civil
df["MARRIAGE"] = df["MARRIAGE"].replace({
    0: 3
})

# Comprobar resultados
print("EDUCATION:")
print(df["EDUCATION"].value_counts().sort_index())

print("\nMARRIAGE:")
print(df["MARRIAGE"].value_counts().sort_index())
```

```python
tasa_educacion = df.groupby("EDUCATION")["DEFAULT"].agg(
    clientes="count",
    incumplimientos="sum",
    tasa_default="mean"
)

tasa_educacion["tasa_default"] *= 100

display(tasa_educacion.round(2))
```

```python
tasa_pay0 = df.groupby("PAY_0")["DEFAULT"].agg(
    clientes="count",
    incumplimientos="sum",
    tasa_default="mean"
)

tasa_pay0["tasa_default"] *= 100

display(tasa_pay0.round(2))
```

```python
import matplotlib.pyplot as plt

tasa_pay0["tasa_default"].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Tasa de incumplimiento según estado de pago")
plt.xlabel("Estado de pago (PAY_0)")
plt.ylabel("Tasa de incumplimiento (%)")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
df["CREDIT_GROUP"] = pd.qcut(
    df["LIMIT_BAL"],
    q=5,
    duplicates="drop"
)

tasa_credito = df.groupby(
    "CREDIT_GROUP",
    observed=True
)["DEFAULT"].agg(
    clientes="count",
    incumplimientos="sum",
    tasa_default="mean"
)

tasa_credito["tasa_default"] *= 100

display(tasa_credito.round(2))
```

Esto podría indicar que los clientes con mayor límite de crédito presentan, en promedio, un perfil de menor riesgo. Sin embargo, no significa que aumentar el límite de crédito reduzca el riesgo: también puede reflejar que el banco ya seleccionó previamente a clientes con mejores perfiles.

```python
df["AGE_GROUP"] = pd.cut(
    df["AGE"],
    bins=[20, 30, 40, 50, 60, 80],
    labels=["20-29", "30-39", "40-49", "50-59", "60+"],
    right=False
)

tasa_edad = df.groupby(
    "AGE_GROUP",
    observed=True
)["DEFAULT"].agg(
    clientes="count",
    incumplimientos="sum",
    tasa_default="mean"
)

tasa_edad["tasa_default"] *= 100

display(tasa_edad.round(2))
```

Observaciones:
- 30-39 años: presentan la menor tasa de incumplimiento (20.25%).
- 60 años o más: presentan la mayor tasa (28.32%), aunque este grupo tiene solamente 339 clientes.
- 50-59 años: también muestran una tasa relativamente elevada (24.86%).

Sin embargo, no debemos concluir que los clientes mayores necesariamente son más riesgosos. Podrían intervenir otras variables.

```python
tasa_sexo = df.groupby("SEX")["DEFAULT"].agg(
    clientes="count",
    incumplimientos="sum",
    tasa_default="mean"
)

tasa_sexo["tasa_default"] *= 100

display(tasa_sexo.round(2))
```

Observamos que:
- Los hombres presentan una tasa de incumplimiento del 24.17%.
- Las mujeres presentan una tasa del 20.78%.
- La diferencia es de 3.39 puntos porcentuales.

```python
tasa_matrimonio = df.groupby("MARRIAGE")["DEFAULT"].agg(
    clientes="count",
    incumplimientos="sum",
    tasa_default="mean"
)

tasa_matrimonio["tasa_default"] *= 100

display(tasa_matrimonio.round(2))
```

Observaciones:
- Los clientes casados presentan una tasa de incumplimiento de 23.47%.
- Los solteros presentan una tasa de 20.93%.
- La categoría otros/desconocido presenta 23.61%, pero tiene una muestra pequeña.
La diferencia entre casados y solteros es de 2.54 puntos porcentuales.

Conclusión preliminar: el estado civil parece tener una asociación descriptiva relativamente débil con el incumplimiento

```python
bill_cols = [
    "BILL_AMT1", "BILL_AMT2", "BILL_AMT3",
    "BILL_AMT4", "BILL_AMT5", "BILL_AMT6"
]

pay_cols = [
    "PAY_AMT1", "PAY_AMT2", "PAY_AMT3",
    "PAY_AMT4", "PAY_AMT5", "PAY_AMT6"
]

# Comparación de montos facturados
display(
    df.groupby("DEFAULT")[bill_cols].mean().round(2)
)

# Comparación de montos pagados
display(
    df.groupby("DEFAULT")[pay_cols].mean().round(2)
)
```

Los clientes que incumplen no necesariamente tienen facturas más elevadas; la diferencia más clara está en los montos que pagan.

```python
import numpy as np
# Proporción pagada respecto a la factura
df["PAYMENT_RATIO"] = np.where(
    df["BILL_AMT1"] > 0,
    df["PAY_AMT1"] / df["BILL_AMT1"],
    np.nan
)

# Utilización del límite de crédito
df["CREDIT_UTILIZATION"] = np.where(
    df["LIMIT_BAL"] > 0,
    df["BILL_AMT1"] / df["LIMIT_BAL"],
    np.nan
)

# Comparación por incumplimiento
display(
    df.groupby("DEFAULT")[
        ["PAYMENT_RATIO", "CREDIT_UTILIZATION"]
    ].mean().round(3)
)
```

Interpretación: los clientes que incumplen parecen tener menor capacidad o disposición para cubrir sus saldos mediante pagos.

Los clientes que incumplen utilizan, en promedio, una proporción mayor de su límite de crédito.

```python
resumen_ratios = df.groupby("DEFAULT")[
    ["PAYMENT_RATIO", "CREDIT_UTILIZATION"]
].agg([
    "count",
    "mean",
    "median",
    "min",
    "max"
])

display(resumen_ratios.round(3))
```

Hay valores extraordinariamente altos que están elevando la media. Por ejemplo, un PAYMENT_RATIO de 11,453 significa que el pago fue más de 11,000 veces el importe facturado. Esto puede ocurrir cuando el importe facturado es muy pequeño.

La utilización del crédito parece ser una variable con potencial predictivo, porque presenta una diferencia importante entre clientes que incumplen y los que no.

En cambio, PAYMENT_RATIO requiere tratamiento adicional debido a sus valores extremos.

```python
# Percentiles de PAYMENT_RATIO
display(
    df.groupby("DEFAULT")["PAYMENT_RATIO"]
    .quantile([0.01, 0.25, 0.50, 0.75, 0.95, 0.99])
    .unstack()
    .round(3)
)

# Percentiles de CREDIT_UTILIZATION
display(
    df.groupby("DEFAULT")["CREDIT_UTILIZATION"]
    .quantile([0.01, 0.25, 0.50, 0.75, 0.95, 0.99])
    .unstack()
    .round(3)
)
```

Interpretación:
- Los clientes que incumplen utilizan una proporción mayor de su crédito disponible.
- La mediana es 0.493 para incumplidos y 0.267 para cumplidos.
- En el percentil 75, ambos grupos presentan una utilización elevada, pero los incumplidos siguen teniendo valores superiores.
Esto sugiere que utilizar una mayor proporción del crédito disponible está asociado con un mayor riesgo de incumplimiento.

```python
print("Variables numéricas:")
print(X_train.select_dtypes(include=["int64", "float64"]).columns.tolist())

print("\nVariables categóricas:")
print(X_train.select_dtypes(include=["object", "category"]).columns.tolist())
```

```python
X = X.drop(columns=["CREDIT_GROUP", "AGE_GROUP"])

variables_categoricas = [
    "SEX",
    "EDUCATION",
    "MARRIAGE"
]

variables_numericas = [
    col for col in X.columns
    if col not in variables_categoricas
]

print("Categóricas:", variables_categoricas)
print("Numéricas:", variables_numericas)
```

```python
# Variables numéricas
procesamiento_numerico = Pipeline([
    ("imputacion", SimpleImputer(strategy="median")),
    ("escalado", StandardScaler())
])

# Variables categóricas
procesamiento_categorico = Pipeline([
    ("imputacion", SimpleImputer(strategy="most_frequent")),
    ("codificacion", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocesamiento = ColumnTransformer([
    ("numerico", procesamiento_numerico, variables_numericas),
    ("categorico", procesamiento_categorico, variables_categoricas)
])
```

```python
from sklearn.linear_model import LogisticRegression

modelo_logistico = Pipeline([
    ("preprocesamiento", preprocesamiento),
    ("modelo", LogisticRegression(
        max_iter=1000,
        class_weight="balanced",
        random_state=42
    ))
])

modelo_logistico.fit(X_train, y_train)
```

```python
predicciones = modelo_logistico.predict(X_test)

probabilidades = modelo_logistico.predict_proba(X_test)[:, 1]
```

```python
# Crear una función para calcular la pérdida esperada
def calcular_perdida(datos):

    return (
        datos['Prob_Default'] *
        datos['LGD'] *
        datos['EAD']
    ).sum()


# Pérdida esperada original
perdida_base = calcular_perdida(resultados)


# ESTRATEGIA 1: CONSERVADORA
conservadora = resultados.copy()

conservadora['EAD'] = conservadora['EAD'].astype(float)

mascara = conservadora['Nivel_Riesgo'].isin(
    ['Alto', 'Muy alto']
)

conservadora.loc[mascara, 'EAD'] *= 0.80

perdida_conservadora = calcular_perdida(conservadora)


# ESTRATEGIA 2: MODERADA
moderada = resultados.copy()

moderada['EAD'] = moderada['EAD'].astype(float)

mascara = moderada['Nivel_Riesgo'] == 'Muy alto'

moderada.loc[mascara, 'EAD'] *= 0.80

perdida_moderada = calcular_perdida(moderada)


# ESTRATEGIA 3: FLEXIBLE
flexible = resultados.copy()

mascara = flexible['Nivel_Riesgo'].isin(
    ['Alto', 'Muy alto']
)

# Reducción relativa del 10% en la PD
flexible.loc[mascara, 'Prob_Default'] *= 0.90

perdida_flexible = calcular_perdida(flexible)


# COMPARACIÓN FINAL
comparacion = pd.DataFrame({
    'Estrategia': [
        'Original',
        'Conservadora',
        'Moderada',
        'Flexible'
    ],
    'Perdida_Esperada': [
        perdida_base,
        perdida_conservadora,
        perdida_moderada,
        perdida_flexible
    ]
})

comparacion['Reduccion_Monetaria'] = (
    perdida_base - comparacion['Perdida_Esperada']
)

comparacion['Reduccion_Porcentual'] = (
    comparacion['Reduccion_Monetaria'] /
    perdida_base * 100
)

comparacion
```
